```python
class BankAccount:

    def __init__(
        self,
        account_number,
        customer_name,
        phone_number,
        initial_deposit
    ):

        self.account_number = account_number
        self.customer_name = customer_name
        self.phone_number = phone_number
        self.balance = initial_deposit
        self.minimum_balance = 1000
        self.account_type = "General Account"

    # Deposit money
    def deposit(self, amount):

        if amount > 0:

            self.balance = self.balance + amount

            print("Amount deposited successfully.")
            print("Updated balance: ₹", self.balance)

        else:

            print("Invalid deposit amount.")

    # Withdraw money
    def withdraw(self, amount):

        if amount <= 0:

            print("Invalid withdrawal amount.")

        elif self.balance - amount < self.minimum_balance:

            print("Withdrawal denied.")
            print(
                "Minimum balance must be maintained: ₹",
                self.minimum_balance
            )

        else:

            self.balance = self.balance - amount

            print("Amount withdrawn successfully.")
            print("Updated balance: ₹", self.balance)

    # Check balance
    def check_balance(self):

        print("Current balance: ₹", self.balance)

    # Display account details
    def display_details(self):

        print("\n----- Account Details -----")
        print("Account Number :", self.account_number)
        print("Customer Name  :", self.customer_name)
        print("Phone Number   :", self.phone_number)
        print("Account Type   :", self.account_type)
        print("Balance        : ₹", self.balance)
        print("Minimum Balance: ₹", self.minimum_balance)
```

```python
class SavingsAccount(BankAccount):

    def __init__(
        self,
        account_number,
        customer_name,
        phone_number,
        initial_deposit
    ):

        super().__init__(
            account_number,
            customer_name,
            phone_number,
            initial_deposit
        )

        self.account_type = "Savings Account"
        self.interest_rate = 4

    # Add interest
    def add_interest(self):

        interest = self.balance * self.interest_rate / 100

        self.balance = self.balance + interest

        print("Interest added successfully.")
        print("Interest amount: ₹", interest)
        print("Updated balance: ₹", self.balance)
```

```python
class CurrentAccount(BankAccount):

    def __init__(
        self,
        account_number,
        customer_name,
        phone_number,
        initial_deposit
    ):

        super().__init__(
            account_number,
            customer_name,
            phone_number,
            initial_deposit
        )

        self.account_type = "Current Account"
        self.overdraft_limit = 5000

    # Override withdrawal method
    def withdraw(self, amount):

        if amount <= 0:

            print("Invalid withdrawal amount.")

        elif self.balance - amount < -self.overdraft_limit:

            print("Withdrawal denied.")
            print(
                "Overdraft limit is: ₹",
                self.overdraft_limit
            )

        else:

            self.balance = self.balance - amount

            print("Amount withdrawn successfully.")
            print("Updated balance: ₹", self.balance)

    # Display overdraft limit
    def display_overdraft_limit(self):

        print("Overdraft Limit: ₹", self.overdraft_limit)
```

```python
class Bank:

    def __init__(self):

        self.accounts = {}
        self.next_account_number = 1001

    # Open a new account
    def open_account(
        self,
        customer_name,
        phone_number,
        initial_deposit,
        account_type
    ):

        if initial_deposit < 1000:

            print("Account cannot be opened.")
            print("Minimum opening deposit is ₹1000.")
            return None

        account_number = self.next_account_number

        if account_type == "Savings":

            account = SavingsAccount(
                account_number,
                customer_name,
                phone_number,
                initial_deposit
            )

        elif account_type == "Current":

            account = CurrentAccount(
                account_number,
                customer_name,
                phone_number,
                initial_deposit
            )

        else:

            print("Invalid account type.")
            return None

        self.accounts[account_number] = account

        self.next_account_number += 1

        print("\nAccount created successfully!")
        print("Account Type:", account.account_type)
        print("Your Account Number is:", account_number)

        return account

    # Find an account
    def find_account(self, account_number):

        if account_number in self.accounts:

            return self.accounts[account_number]

        else:

            print("Account not found.")
            return None
```

```python
# ---------- VALIDATION FUNCTIONS ----------

def get_name():

    while True:
        name = input("Enter customer name: ").strip()

        if name == "":
            print("Name cannot be empty.")

        elif not name.replace(" ", "").isalpha():
            print("Name should contain only letters.")

        else:
            return name


def get_phone():

    while True:
        phone = input("Enter phone number: ").strip()

        if not phone.isdigit():
            print("Phone number must contain only digits.")

        elif len(phone) != 10:
            print("Phone number must contain exactly 10 digits.")

        elif phone[0] not in "6789":
            print("Phone number must start with 6, 7, 8, or 9.")

        else:
            return phone

def get_account_number():

    while True:
        try:
            account_number = int(
                input("Enter account number: ")
            )

            if account_number <= 0:
                print("Account number must be positive.")

            else:
                return account_number

        except ValueError:
            print("Enter a valid account number.")


def get_positive_amount(message):

    while True:
        try:
            amount = float(input(message))

            if amount <= 0:
                print("Amount must be greater than ₹0.")

            else:
                return amount

        except ValueError:
            print("Enter a valid numeric amount.")


def get_initial_deposit():

    while True:
        try:
            amount = float(
                input("Enter initial deposit: ")
            )

            if amount < 1000:
                print("Minimum opening deposit is ₹1000.")

            else:
                return amount

        except ValueError:
            print("Enter a valid numeric amount.")
```

```python
# ---------- MENU-DRIVEN BANKING SYSTEM ----------

while True:

    print("\n==============================")
    print("       BANKING SYSTEM")
    print("==============================")
    print("1. Open New Account")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Check Balance")
    print("5. Account Details")
    print("6. Add Interest (Savings)")
    print("7. Check Overdraft Limit (Current)")
    print("8. Exit")
    print("==============================")

    choice = input("Enter your choice: ").strip()

    # ---------- OPEN ACCOUNT ----------

    if choice == "1":

        name = get_name()

        phone = get_phone()

        amount = get_initial_deposit()

        # Account type validation
        while True:

            print("\nSelect Account Type:")
            print("1. Savings Account")
            print("2. Current Account")

            account_choice = input(
                "Enter account type: "
            ).strip()

            if account_choice == "1":
                account_type = "Savings"
                break

            elif account_choice == "2":
                account_type = "Current"
                break

            else:
                print("Invalid account type. Choose 1 or 2.")

        bank.open_account(
            name,
            phone,
            amount,
            account_type
        )

    # ---------- DEPOSIT ----------

    elif choice == "2":

        account_number = get_account_number()

        amount = get_positive_amount(
            "Enter deposit amount: "
        )

        account = bank.find_account(account_number)

        if account:
            account.deposit(amount)

    # ---------- WITHDRAW ----------

    elif choice == "3":

        account_number = get_account_number()

        amount = get_positive_amount(
            "Enter withdrawal amount: "
        )

        account = bank.find_account(account_number)

        if account:
            account.withdraw(amount)

    # ---------- CHECK BALANCE ----------

    elif choice == "4":

        account_number = get_account_number()

        account = bank.find_account(account_number)

        if account:
            account.check_balance()

    # ---------- ACCOUNT DETAILS ----------

    elif choice == "5":

        account_number = get_account_number()

        account = bank.find_account(account_number)

        if account:
            account.display_details()

    # ---------- ADD INTEREST ----------

    elif choice == "6":

        account_number = get_account_number()

        account = bank.find_account(account_number)

        if account:

            if isinstance(account, SavingsAccount):
                account.add_interest()

            else:
                print("This is not a Savings Account.")

    # ---------- OVERDRAFT LIMIT ----------

    elif choice == "7":

        account_number = get_account_number()

        account = bank.find_account(account_number)

        if account:

            if isinstance(account, CurrentAccount):
                account.display_overdraft_limit()

            else:
                print("This is not a Current Account.")

    # ---------- EXIT ----------

    elif choice == "8":

        print("Thank you for using our Banking System!")
        break

    # ---------- INVALID CHOICE ----------

    else:

        print("Invalid choice. Please select 1 to 8.")
```
